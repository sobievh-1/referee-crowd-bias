```python
# Tiny fake example in the shape of our future data
matches = pd.DataFrame({
    'HomeTeam': ['Arsenal', 'Chelsea', 'Everton'],
    'AwayTeam': ['Spurs', 'Leeds', 'Wolves'],
    'HY': [1, 2, 3],
    'AY': [3, 2, 4],
})
matches['card_diff'] = matches['AY'] - matches['HY']
matches
```

```python
fig, ax = plt.subplots(figsize=(4, 3))
ax.bar(matches['HomeTeam'], matches['card_diff'])
ax.set_ylabel('card_diff (away - home)')
plt.show()
```
